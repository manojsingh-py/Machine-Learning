```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv('../datasets/covid_toy.csv')
```

```python
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import OrdinalEncoder
```

```python
df.isna().sum()
```

```python
df['cough'].value_counts()
```

```python
from sklearn.model_selection import train_test_split
X = df.drop('has_covid', axis=1)
y = df['has_covid']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)
```

```python
X_train
```

# 1. Aam Zindagi

```python
si = SimpleImputer()

# adding simple imputer to fever col
X_train_fever = si.fit_transform(X_train[['fever']])

# also the test datasets
X_test_fever = si.transform(X_test[['fever']])
```

```python
X_train_fever
```

```python
# Ordinalencoding -> cough

oe = OrdinalEncoder(categories=[['Mild', 'Strong']])

X_train_cough = oe.fit_transform(X_train[['cough']])
X_test_cough = oe.transform(X_test[['cough']])
```

```python
X_train_cough
```

```python
# OneHotEncoding -> gender,city

ohe = OneHotEncoder(drop='first', sparse_output=False)

X_train_gender = ohe.fit_transform(X_train[['gender']])
X_test_gender = ohe.transform(X_test[['gender']])
```

```python
X_train_gender
```

```python
X_train_city = ohe.fit_transform(X_train[['city']])
X_test_city = ohe.transform(X_test[['city']])
```

```python
X_train_city
```

```python
# Extracting Age
X_train_age = X_train[['age']]
X_test_age = X_test[['age']]

# # Extracting Age
# X_train_age = X_train.drop(columns=['gender','fever','cough','city']).values
#
# # also the test datasets
# X_test_age = X_test.drop(columns=['gender','fever','cough','city']).values
```

# Mentos Zindagi

```python
from sklearn.compose import ColumnTransformer
```

```python
transformer = ColumnTransformer(transformers=[
    ('tnf1', SimpleImputer(), ['fever']),
    ('tnf2', OrdinalEncoder(categories=[['Mild', 'Strong']]), ['cough']),
    ('tnf3', OneHotEncoder(drop='first', sparse_output=False), ['gender', 'city']),
], remainder='passthrough')
```

```python
transformer.fit(X_train)
```

```python
X_train_trns = transformer.fit_transform(X_train)
```

```python
X_train_trns, X_train_trns.shape
```

```python
X_test_trns = transformer.transform(X_test)
```
