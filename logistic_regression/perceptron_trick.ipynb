```python
from matplotlib.pyplot import step
from sklearn.datasets import make_classification
import numpy as np

import matplotlib.pyplot as plt
```

```python
X, y = make_classification(n_samples=100, n_features=2, n_informative=1,n_redundant=0,
                           n_classes=2, n_clusters_per_class=1, random_state=41,hypercube=False,class_sep=10)
```

```python
plt.figure(figsize=(10,6))
plt.scatter(X[:,0],X[:,1],c=y,cmap='winter',s=100)
```

```python
def perceptron_trick(X,y):
    X = np.insert(X,0,1,axis=1)
    weight = np.zeros(X.shape[1])
    learning_rate = 0.1

    for _ in range(100):
        j = np.random.randint(0, 100)
        y_hat = step(np.dot(X[j],weight))
        weight = weight + learning_rate * (y[j] - y_hat)*X[j]

    return weight[0], weight[1:]
```

```python
def step(x):
    if x >= 0:
        return 1
    else:
        return 0
```

```python
intercept_, coef_ = perceptron_trick(X,y)

print('intercept_:',intercept_)
print('coef_:',coef_)
```

```python
m = -(coef_[0]/coef_[1])
b = -(intercept_/coef_[1])
```

```python
x_input = np.linspace(-3,3,100)
y_input = m*x_input + b
```

```python
plt.figure(figsize=(10,6))
plt.plot(x_input,y_input,color='red',linewidth=3)
plt.scatter(X[:,0],X[:,1],c=y,cmap='winter',s=100)
plt.ylim(-3,2)
```

```python
def perceptron(X,y):

    m = []
    b = []

    X = np.insert(X,0,1,axis=1)
    weights = np.ones(X.shape[1])
    lr = 0.1

    for i in range(200):
        j = np.random.randint(0,100)
        y_hat = step(np.dot(X[j],weights))
        weights = weights + lr*(y[j]-y_hat)*X[j]

        m.append(-(weights[1]/weights[2]))
        b.append(-(weights[0]/weights[2]))

    return m,b
```

```python
m,b = perceptron(X,y)
```

```python
plt.figure(figsize=(10,6))
for i in range(len(m)):
    y_input = m[i]*x_input + b[i]
    plt.plot(x_input,y_input,color='red',alpha=0.1)
plt.scatter(X[:,0],X[:,1],c=y,cmap='winter',s=100)
plt.ylim(-3,2)
```
